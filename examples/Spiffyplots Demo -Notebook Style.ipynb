# Spiffyplots notebook style demo

This notebook compares the base `spiffy` style with the larger `notebook` text modifier for inline figures.

## 1. Style comparison

How does the notebook modifier change text scale without changing the canvas? The same seeded example plots are drawn with `spiffy` and with `["spiffy", "notebook"]`.

```python
import matplotlib.pyplot as plt
from IPython.display import display

from spiffyplots import MultiPanel, populate_random_data

SEED = 7
PANEL_SHAPE = (1, 2)
STYLE_STACKS = ("spiffy", ["spiffy", "notebook"])
SUPTITLES = ("Style: 'spiffy'", "Style: 'spiffy' with 'notebook'")
```

```python
style_cases = tuple(zip(STYLE_STACKS, SUPTITLES, strict=True))
```

```python
for styles, suptitle in style_cases:
    with plt.style.context(styles):
        figure = MultiPanel(shape=PANEL_SHAPE)
        populate_random_data(figure.panels, seed=SEED)
        figure.fig.suptitle(suptitle)
        display(figure.fig)
        figure.close()
```
